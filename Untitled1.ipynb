```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
df.head()
```

```python
print(df.shape)
df.info()
print(df.isnull().sum())
print("Duplicates:", df.duplicated().sum())
```

```python
df = pd.concat([df, df.head(10)])
print("Duplicate rows now:", df.duplicated().sum())
```

```python
df = df.drop(columns=["Cabin"])
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
print(df.isnull().sum())
```

```python
print("Before:", df.shape)
df = df.drop_duplicates()
print("After:", df.shape)
```

```python
df.columns = df.columns.str.lower()
df["sex"] = df["sex"].str.strip().str.lower()
df["embarked"] = df["embarked"].str.strip().str.upper()
df["age"] = df["age"].astype(int)
df.info()
```

```python
clean_data = df.copy()
model_data = df.drop(columns=["passengerid", "name", "ticket"])
model_data = pd.get_dummies(model_data, columns=["sex", "embarked"], drop_first=True)

model_data.head()
```

```python
model_data.to_csv("cleaned_data.csv", index=False)

from google.colab import files
files.download("cleaned_data.csv")
```

```python
import matplotlib.pyplot as plt

clean_data["age"].hist(bins=20)
plt.title("Age of passengers")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()

clean_data["fare"].hist(bins=20)
plt.title("Ticket fare")
plt.xlabel("Fare")
plt.ylabel("Count")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(x="survived", data=clean_data)
plt.title("Survived or not")
plt.show()

sns.countplot(x="sex", hue="survived", data=clean_data)
plt.title("Survival by gender")
plt.show()

sns.countplot(x="pclass", hue="survived", data=clean_data)
plt.title("Survival by class")
plt.show()
```

```python
numbers = clean_data[["survived", "pclass", "age", "sibsp", "parch", "fare"]]

plt.figure(figsize=(7, 5))
sns.heatmap(numbers.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation plot")
plt.show()
```
