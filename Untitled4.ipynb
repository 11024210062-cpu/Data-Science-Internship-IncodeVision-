```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
data = {
    'Age': [20, 22, 25, 30, 35, 40, 45, 50, 55, 60,
            21, 24, 28, 32, 38],

    'Annual_Income': [15, 18, 20, 25, 30, 35, 40, 45,
                      50, 55, 16, 22, 28, 32, 38],

    'Spending_Score': [80, 85, 75, 60, 65, 40, 45, 30,
                       25, 20, 90, 78, 62, 55, 42]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.head())
print(df.isnull().sum())
print(df.info())
```

```python
X = df[['Age', 'Annual_Income', 'Spending_Score']]
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled)
```

```python
model = KMeans(n_clusters=3, random_state=42, n_init=10)
df['Customer_Group'] = model.fit_predict(X_scaled)
print(df)
```

```python
plt.scatter(
    df['Annual_Income'],
    df['Spending_Score'],
    c=df['Customer_Group'])
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Customer Segmentation")
plt.show()
```

```python
result = df.groupby('Customer_Group').mean()
print(result)
```
